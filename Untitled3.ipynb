```python
!pip install dgeb datasets -q

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset, WeightedRandomSampler
from sklearn.metrics import average_precision_score, classification_report
from sklearn.model_selection import train_test_split
from itertools import product
from datasets import load_dataset

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {DEVICE}")

DATASET_NAMES = {
    "ecoli":  ("tattabio/ecoli_operonic_pair",  "a62c01143a842696fc8200b91c1acb825e8cb891"),
    "cyano":  ("tattabio/cyano_operonic_pair",  "eeb4cb71ec2a4ff688af9de7c0662123577d32ec"),
    "vibrio": ("tattabio/vibrio_operonic_pair", "24781b12b45bf81a079a6164ef0d2124948c1878"),
}

AA = "ACDEFGHIKLMNPQRSTVWY"
KMERS = {"".join(p): i for i, p in enumerate(product(AA, repeat=3))}
KMER_DIM = len(KMERS)
PAIR_DIM = KMER_DIM * 3

BATCH_SIZE = 32
LR = 5e-4
EPOCHS = 80
PATIENCE = 12
WEIGHT_DECAY = 1e-3


def load_operon_data(organism):
    path, revision = DATASET_NAMES[organism]
    ds = load_dataset(path, revision=revision)
    split = ds["train"] if "train" in ds else ds[list(ds.keys())[0]]
    seqs = split["Sequence"]
    labels = split["Label"]
    seq_a, seq_b, pair_labels = [], [], []
    for i in range(len(seqs) - 1):
        seq_a.append(seqs[i])
        seq_b.append(seqs[i + 1])
        pair_labels.append(int(labels[i]))
    pair_labels = np.array(pair_labels)
    pos = pair_labels.sum()
    print(f"  [{organism}] {len(pair_labels)} pairs | {pos} same-operon ({pair_labels.mean():.1%}) | {len(pair_labels)-pos} different-operon")
    return seq_a, seq_b, pair_labels


def kmer_vector(seq):
    vec = np.zeros(KMER_DIM, dtype=np.float32)
    seq = seq.upper()
    count = 0
    for i in range(len(seq) - 2):
        kmer = seq[i:i+3]
        if kmer in KMERS:
            vec[KMERS[kmer]] += 1
            count += 1
    if count > 0:
        vec /= count
    return vec


def encode_pairs(seq_a_list, seq_b_list):
    vA = np.array([kmer_vector(s) for s in seq_a_list])
    vB = np.array([kmer_vector(s) for s in seq_b_list])
    return np.concatenate([vA, vB, np.abs(vA - vB)], axis=1).astype(np.float32)


class ResidualBlock1D(nn.Module):
    def __init__(self, channels, kernel_size=3):
        super().__init__()
        self.conv1 = nn.Conv1d(channels, channels, kernel_size, padding=kernel_size // 2)
        self.bn1 = nn.BatchNorm1d(channels)
        self.conv2 = nn.Conv1d(channels, channels, kernel_size, padding=kernel_size // 2)
        self.bn2 = nn.BatchNorm1d(channels)

    def forward(self, x):
        residual = x
        x = F.relu(self.bn1(self.conv1(x)))
        x = self.bn2(self.conv2(x))
        return F.relu(x + residual)


class KmerCNN(nn.Module):
    def __init__(self, input_dim=PAIR_DIM, num_channels=64, num_residual=2):
        super().__init__()
        self.proj = nn.Sequential(
            nn.Linear(input_dim, 1024),
            nn.BatchNorm1d(1024),
            nn.ReLU(),
            nn.Dropout(0.4),
        )
        self.conv_in = nn.Sequential(
            nn.Conv1d(1, num_channels, kernel_size=7, padding=3),
            nn.BatchNorm1d(num_channels),
            nn.ReLU(),
        )
        self.res_blocks = nn.Sequential(
            *[ResidualBlock1D(num_channels) for _ in range(num_residual)]
        )
        self.pool = nn.AdaptiveAvgPool1d(8)
        self.head = nn.Sequential(
            nn.Flatten(),
            nn.Linear(num_channels * 8, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(256, 1),
        )

    def forward(self, x):
        x = self.proj(x)
        x = x.unsqueeze(1)
        x = self.conv_in(x)
        x = self.res_blocks(x)
        x = self.pool(x)
        return self.head(x).squeeze(1)


def get_sampler(labels):
    counts = np.bincount(labels)
    weights = 1.0 / counts[labels]
    return WeightedRandomSampler(torch.FloatTensor(weights), len(weights), replacement=True)


def predict_batched(model, X, batch_size=128):
    model.eval()
    all_probs = []
    with torch.no_grad():
        for start in range(0, len(X), batch_size):
            xb = torch.FloatTensor(X[start:start+batch_size]).to(DEVICE)
            all_probs.append(torch.sigmoid(model(xb)).cpu().numpy())
    return np.concatenate(all_probs)


def train_model(X_train, y_train, X_val, y_val, pos_rate):
    pos_count = y_train.sum()
    neg_count = len(y_train) - pos_count

    # Aggressive pos_weight for heavily imbalanced datasets
    pw_ratio = neg_count / pos_count
    if pos_rate < 0.35:
        pw_ratio *= 2.0
    pos_weight = torch.tensor([pw_ratio], dtype=torch.float32).to(DEVICE)
    print(f"  pos_weight: {pw_ratio:.2f} (pos_rate={pos_rate:.1%})")

    sampler = get_sampler(y_train.astype(int))
    loader = DataLoader(
        TensorDataset(torch.FloatTensor(X_train), torch.FloatTensor(y_train)),
        batch_size=BATCH_SIZE, sampler=sampler, drop_last=True
    )

    model = KmerCNN().to(DEVICE)
    optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
    criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

    best_val_ap = 0
    best_state = None
    patience_counter = 0

    for epoch in range(EPOCHS):
        model.train()
        epoch_loss = 0
        for xb, yb in loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)
            optimizer.zero_grad()
            loss = criterion(model(xb), yb)
            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            epoch_loss += loss.item()
        scheduler.step()

        val_probs = predict_batched(model, X_val)
        val_ap = average_precision_score(y_val, val_probs)
        print(f"  Epoch {epoch+1:3d} | loss: {epoch_loss/len(loader):.4f} | val AP: {val_ap:.4f}")

        if val_ap > best_val_ap:
            best_val_ap = val_ap
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1

        if patience_counter >= PATIENCE:
            print(f"  Early stopping at epoch {epoch+1} | best val AP: {best_val_ap:.4f}")
            break

    model.load_state_dict(best_state)
    return model


def train_and_evaluate(organism, seq_a, seq_b, labels):
    print(f"\nEncoding {organism} pairs...")
    X = encode_pairs(seq_a, seq_b)
    pos_rate = labels.mean()

    X_train, X_test, y_train, y_test = train_test_split(
        X, labels, test_size=0.2, stratify=labels, random_state=42
    )
    X_train, X_val, y_train, y_val = train_test_split(
        X_train, y_train, test_size=0.1, stratify=y_train, random_state=42
    )
    print(f"  Train: {len(X_train)} | Val: {len(X_val)} | Test: {len(X_test)}")

    print(f"Training CNN on {organism}...")
    model = train_model(X_train, y_train, X_val, y_val, pos_rate)

    probs = predict_batched(model, X_test)
    ap = average_precision_score(y_test, probs)

    # Use prior probability as threshold instead of fixed 0.5
    threshold = pos_rate
    preds = (probs >= threshold).astype(int)

    print(f"\n{'─'*50}")
    print(f"[CNN+Kmer | {organism}] Average Precision: {ap:.4f}  (threshold={threshold:.2f})")
    print(classification_report(y_test, preds, target_names=["diff_operon", "same_operon"]))

    return round(ap, 4)


print("Loading datasets...")
sa_ec, sb_ec, y_ec = load_operon_data("ecoli")
sa_cy, sb_cy, y_cy = load_operon_data("cyano")
sa_vi, sb_vi, y_vi = load_operon_data("vibrio")

print("2.6  Neural CNN on K-mer Features")

cnn_results = {}
cnn_results["E. coli"] = train_and_evaluate("E. coli", sa_ec, sb_ec, y_ec)
cnn_results["Cyano"]   = train_and_evaluate("Cyano",   sa_cy, sb_cy, y_cy)
cnn_results["Vibrio"]  = train_and_evaluate("Vibrio",  sa_vi, sb_vi, y_vi)

dgeb_baselines = {
    "E. coli": {"ESM2-650M": 0.5727, "ESM3": 0.5774, "ProtTrans": 0.6180},
    "Cyano":   {"ESM2-650M": 0.3733, "ESM3": 0.3899, "ProtTrans": 0.4092},
    "Vibrio":  {"ESM2-650M": 0.4828, "ESM3": 0.4971, "ProtTrans": 0.5427},
}

print("\n" + "="*60)
print("SUMMARY — Average Precision")
print(f"{'Model':<25} {'E.coli':>10} {'Cyano':>10} {'Vibrio':>10}")
print(f"{'CNN+Kmer (ours)':<25} {cnn_results['E. coli']:>10.4f} {cnn_results['Cyano']:>10.4f} {cnn_results['Vibrio']:>10.4f}")
for model_name in ["ESM2-650M", "ESM3", "ProtTrans"]:
    ec = dgeb_baselines["E. coli"][model_name]
    cy = dgeb_baselines["Cyano"][model_name]
    vi = dgeb_baselines["Vibrio"][model_name]
    print(f"  {model_name:<23} {ec:>10.4f} {cy:>10.4f} {vi:>10.4f}")
print("  (DGEB baselines = zero-shot cosine similarity, no task training)")
```
